```python
import cv2 as cv
from pathlib import Path
```

```python
def show_icon(icon):
    cv.imshow('icon', icon)
    cv.waitKey(0)
    cv.destroyAllWindows()

def resize_icon(icon, size, save_path=None):
    new_icon = cv.resize(icon, size)
    if save_path:
        cv.imwrite(str(save_path), new_icon)
    return new_icon
```

```python
color_root = Path('Color/icon')
color_root.glob('*.png')

for color_path in color_root.glob('*.png'):
    icon = cv.imread(str(color_path), cv.IMREAD_UNCHANGED)
    icon_size = icon.shape[:2]
    size = (128, 128)
    icon_path = color_path.parent.parent / 'mini' / color_path.name.replace(f'_{icon_size[0]}', f'_{size[0]}')
    resize_icon(icon, size, icon_path)
    print(f'resize png {color_path.name} from {icon_size} to {size}')
```
